```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))
variaveis = [
    ("age", "Idade", "#4C78A8"),
    ("weight", "Peso", "#F58518"),
    ("height", "Altura", "#54A24B"),
]

for ax, (coluna, nome, cor) in zip(axes, variaveis):
    sns.histplot(
        data=df,
        x=coluna,
        bins=30,
        kde=True,
        color=cor,
        alpha=0.78,
        edgecolor="white",
        linewidth=0.5,
        ax=ax,
    )
    ax.set_title(nome, fontweight="bold")
    ax.set_xlabel({"age": "Idade (anos)", "weight": "Peso (kg)", "height": "Altura (cm)"}[coluna])
    ax.set_ylabel("Frequência")
    ax.grid(axis="y", alpha=0.25)
    ax.grid(axis="x", visible=False)

fig.suptitle("Distribuição das variáveis", fontsize=16, fontweight="bold", y=1.02)
plt.tight_layout()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8), sharex=True)
relacoes = [
    ("weight", "Peso (kg)", "Idade vs. peso", "#F58518"),
    ("height", "Altura (cm)", "Idade vs. altura", "#54A24B"),
]

for ax, (coluna, rotulo_y, titulo, cor) in zip(axes, relacoes):
    sns.scatterplot(
        data=df,
        x="age",
        y=coluna,
        color=cor,
        alpha=0.35,
        s=32,
        edgecolor="white",
        linewidth=0.25,
        ax=ax,
    )
    sns.regplot(
        data=df,
        x="age",
        y=coluna,
        scatter=False,
        color="#333333",
        line_kws={"linewidth": 2, "linestyle": "--"},
        ax=ax,
    )
    ax.set_title(titulo, fontweight="bold")
    ax.set_xlabel("Idade (anos)")
    ax.set_ylabel(rotulo_y)
    ax.grid(alpha=0.25)

fig.suptitle("Relação entre idade e medidas corporais", fontsize=16, fontweight="bold", y=1.02)
plt.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="age",
    y="height",
    color="#ff4040",
    alpha=0.35,
    s=36,
    edgecolor="white",
    linewidth=0.25,
    ax=ax,
)
sns.regplot(
    data=df,
    x="age",
    y="height",
    scatter=False,
    color="#333333",
    line_kws={"linewidth": 2, "linestyle": "--"},
    ax=ax,
)
ax.set_title("Idade x Altura", fontsize=16, pad=14)
ax.set_xlabel("Idade (anos)")
ax.set_ylabel("Altura (cm)")
ax.grid(alpha=0.25)
plt.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="age",
    y="height",
    color="#A8D5A2",
    showfliers=False,
    width=0.65,
    ax=ax,
)

sns.stripplot(
    data=df,
    x="age",
    y="height",
    color="#2F6F3E",
    alpha=0.12,
    jitter=0.25,
    size=2.5,
    ax=ax,
)

ax.set_title("Distribuição da altura por idade", fontsize=16, fontweight="bold")
ax.set_xlabel("Idade (anos)")
ax.set_ylabel("Altura (cm)")
ax.grid(axis="y", alpha=0.25)
ax.grid(axis="x", visible=False)

plt.tight_layout()
```
